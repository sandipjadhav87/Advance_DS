```python
# ============================================================
# HANDWRITTEN DIGIT RECOGNITION USING CNN
# Dataset: MNIST
# Subject: Advanced Data Science (ADS)
# ============================================================

# -----------------------------
# 1. IMPORT LIBRARIES
# -----------------------------

import os
import numpy as np
import matplotlib.pyplot as plt
import seaborn as sns

from tensorflow import keras
from tensorflow.keras import layers
from tensorflow.keras.models import Sequential
from tensorflow.keras.utils import to_categorical

from sklearn.metrics import (
    confusion_matrix,
    classification_report
)

# Create output directories
os.makedirs("outputs", exist_ok=True)
os.makedirs("saved_model", exist_ok=True)


# ============================================================
# 2. LOAD MNIST DATASET
# ============================================================

print("\n" + "=" * 60)
print("LOADING MNIST DATASET")
print("=" * 60)

(x_train, y_train), (x_test, y_test) = keras.datasets.mnist.load_data()

print("Training images :", x_train.shape)
print("Training labels :", y_train.shape)
print("Testing images  :", x_test.shape)
print("Testing labels  :", y_test.shape)


# ============================================================
# 3. DISPLAY SAMPLE IMAGES
# ============================================================

plt.figure(figsize=(10, 6))

for i in range(15):
    plt.subplot(3, 5, i + 1)
    plt.imshow(x_train[i], cmap="gray")
    plt.title(f"Label: {y_train[i]}")
    plt.axis("off")

plt.suptitle("Sample Images from MNIST Dataset")
plt.tight_layout()

plt.savefig(
    "outputs/sample_images.png",
    dpi=300,
    bbox_inches="tight"
)

plt.show()


# ============================================================
# 4. DATA PREPROCESSING
# ============================================================

print("\n" + "=" * 60)
print("DATA PREPROCESSING")
print("=" * 60)

# Normalize pixel values from [0,255] to [0,1]
x_train = x_train.astype("float32") / 255.0
x_test = x_test.astype("float32") / 255.0

# CNN expects:
# (samples, height, width, channels)

x_train = np.expand_dims(x_train, axis=-1)
x_test = np.expand_dims(x_test, axis=-1)

print("After preprocessing:")
print("Training shape:", x_train.shape)
print("Testing shape :", x_test.shape)


# ============================================================
# 5. ONE-HOT ENCODE LABELS
# ============================================================

num_classes = 10

y_train_cat = to_categorical(
    y_train,
    num_classes
)

y_test_cat = to_categorical(
    y_test,
    num_classes
)

print("Encoded label shape:", y_train_cat.shape)


# ============================================================
# 6. BUILD CNN MODEL
# ============================================================

print("\n" + "=" * 60)
print("BUILDING CNN MODEL")
print("=" * 60)

model = Sequential([

    # Input Layer
    layers.Input(shape=(28, 28, 1)),

    # Convolution Layer 1
    layers.Conv2D(
        filters=32,
        kernel_size=(3, 3),
        activation="relu"
    ),

    # Pooling Layer 1
    layers.MaxPooling2D(
        pool_size=(2, 2)
    ),

    # Convolution Layer 2
    layers.Conv2D(
        filters=64,
        kernel_size=(3, 3),
        activation="relu"
    ),

    # Pooling Layer 2
    layers.MaxPooling2D(
        pool_size=(2, 2)
    ),

    # Flatten
    layers.Flatten(),

    # Fully Connected Layer
    layers.Dense(
        128,
        activation="relu"
    ),

    # Dropout
    layers.Dropout(0.5),

    # Output Layer
    layers.Dense(
        num_classes,
        activation="softmax"
    )
])


# ============================================================
# 7. DISPLAY MODEL SUMMARY
# ============================================================

model.summary()


# ============================================================
# 8. COMPILE MODEL
# ============================================================

model.compile(
    optimizer="adam",
    loss="categorical_crossentropy",
    metrics=["accuracy"]
)


# ============================================================
# 9. TRAIN CNN MODEL
# ============================================================

print("\n" + "=" * 60)
print("TRAINING CNN MODEL")
print("=" * 60)

history = model.fit(
    x_train,
    y_train_cat,

    validation_split=0.1,

    epochs=10,

    batch_size=128,

    verbose=1
)


# ============================================================
# 10. SAVE MODEL
# ============================================================

model_path = "saved_model/mnist_cnn.keras"

model.save(model_path)

print("\nModel saved successfully!")
print("Location:", model_path)


# ============================================================
# 11. PLOT TRAINING ACCURACY
# ============================================================

plt.figure(figsize=(10, 6))

plt.plot(
    history.history["accuracy"],
    label="Training Accuracy"
)

plt.plot(
    history.history["val_accuracy"],
    label="Validation Accuracy"
)

plt.title("CNN Training and Validation Accuracy")

plt.xlabel("Epoch")

plt.ylabel("Accuracy")

plt.legend()

plt.grid(True)

plt.savefig(
    "outputs/training_accuracy.png",
    dpi=300,
    bbox_inches="tight"
)

plt.show()


# ============================================================
# 12. PLOT TRAINING LOSS
# ============================================================

plt.figure(figsize=(10, 6))

plt.plot(
    history.history["loss"],
    label="Training Loss"
)

plt.plot(
    history.history["val_loss"],
    label="Validation Loss"
)

plt.title("CNN Training and Validation Loss")

plt.xlabel("Epoch")

plt.ylabel("Loss")

plt.legend()

plt.grid(True)

plt.savefig(
    "outputs/training_loss.png",
    dpi=300,
    bbox_inches="tight"
)

plt.show()


# ============================================================
# 13. EVALUATE MODEL ON TEST DATA
# ============================================================

print("\n" + "=" * 60)
print("MODEL EVALUATION")
print("=" * 60)

test_loss, test_accuracy = model.evaluate(
    x_test,
    y_test_cat,
    verbose=0
)

print(f"\nTest Loss     : {test_loss:.4f}")
print(f"Test Accuracy : {test_accuracy * 100:.2f}%")


# ============================================================
# 14. MAKE PREDICTIONS
# ============================================================

print("\n" + "=" * 60)
print("GENERATING PREDICTIONS")
print("=" * 60)

predictions = model.predict(
    x_test,
    verbose=0
)

predicted_classes = np.argmax(
    predictions,
    axis=1
)


# ============================================================
# 15. CLASSIFICATION REPORT
# ============================================================

print("\n" + "=" * 60)
print("CLASSIFICATION REPORT")
print("=" * 60)

print(
    classification_report(
        y_test,
        predicted_classes,
        digits=4
    )
)


# ============================================================
# 16. CONFUSION MATRIX
# ============================================================

cm = confusion_matrix(
    y_test,
    predicted_classes
)

plt.figure(figsize=(10, 8))

sns.heatmap(
    cm,
    annot=True,
    fmt="d",
    cmap="Blues",
    xticklabels=range(10),
    yticklabels=range(10)
)

plt.title("Confusion Matrix - MNIST CNN")

plt.xlabel("Predicted Label")

plt.ylabel("Actual Label")

plt.savefig(
    "outputs/confusion_matrix.png",
    dpi=300,
    bbox_inches="tight"
)

plt.show()


# ============================================================
# 17. DISPLAY PREDICTIONS
# ============================================================

plt.figure(figsize=(12, 8))

for i in range(20):

    plt.subplot(4, 5, i + 1)

    plt.imshow(
        x_test[i].reshape(28, 28),
        cmap="gray"
    )

    actual = y_test[i]
    predicted = predicted_classes[i]

    if actual == predicted:
        color = "green"
    else:
        color = "red"

    plt.title(
        f"Actual: {actual}\nPredicted: {predicted}",
        color=color
    )

    plt.axis("off")

plt.suptitle(
    "CNN Predictions on MNIST Test Images"
)

plt.tight_layout()

plt.savefig(
    "outputs/predictions.png",
    dpi=300,
    bbox_inches="tight"
)

plt.show()


# ============================================================
# 18. TEST A SINGLE IMAGE
# ============================================================

print("\n" + "=" * 60)
print("SINGLE IMAGE PREDICTION")
print("=" * 60)

index = 100

single_image = x_test[index]

single_image_batch = np.expand_dims(
    single_image,
    axis=0
)

prediction = model.predict(
    single_image_batch,
    verbose=0
)

predicted_digit = np.argmax(
    prediction
)

confidence = np.max(
    prediction
) * 100

print("Actual Digit    :", y_test[index])
print("Predicted Digit :", predicted_digit)
print(f"Confidence      : {confidence:.2f}%")


# ============================================================
# 19. DISPLAY SINGLE IMAGE
# ============================================================

plt.figure(figsize=(5, 5))

plt.imshow(
    single_image.reshape(28, 28),
    cmap="gray"
)

plt.title(
    f"Predicted Digit: {predicted_digit}\n"
    f"Confidence: {confidence:.2f}%"
)

plt.axis("off")

plt.show()


# ============================================================
# 20. FINAL RESULT
# ============================================================

print("\n" + "=" * 60)
print("FINAL RESULT")
print("=" * 60)

print(
    f"CNN Test Accuracy: {test_accuracy * 100:.2f}%"
)

print(
    "\nHandwritten digit recognition completed successfully!"
)

print(
    "\nGenerated files:"
)

print("1. outputs/sample_images.png")
print("2. outputs/training_accuracy.png")
print("3. outputs/training_loss.png")
print("4. outputs/confusion_matrix.png")
print("5. outputs/predictions.png")
print("6. saved_model/mnist_cnn.keras")
```
